```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder

df = pd.DataFrame({
    'age': [25, None, 40, 35, None],
    'income': [30000, 45000, None, 80000, 52000],
    'city': ['Chennai', 'Mumbai', None, 'Chennai', 'Delhi'],
    'size': ['S', 'L', 'M', 'S', 'L']
})

print(df)
```

```python
print("Missing Values Before Imputation:")
print(df.isnull().sum().to_string())
```

```python
age_imputer = SimpleImputer(strategy='median')

df['age'] = age_imputer.fit_transform(df[['age']])

print("DataFrame After Age Imputation:")
print(df)
```

```python
income_imputer = SimpleImputer(strategy='median')

df['income'] = income_imputer.fit_transform(df[['income']])

print("DataFrame After Income Imputation:")
print(df)
```

```python
city_imputer = SimpleImputer(strategy='most_frequent')

df['city'] = city_imputer.fit_transform(df[['city']]).ravel()

print("DataFrame After City Imputation:")
print(df)
```

```python
df_knn = pd.DataFrame({
    'age': [25, None, 40, 35, None],
    'income': [30000, 45000, None, 80000, 52000]
})

knn_imputer = KNNImputer(n_neighbors=2)

df_knn[['age', 'income']] = knn_imputer.fit_transform(
    df_knn[['age', 'income']]
)

print("DataFrame After KNN Imputation:")
print(df_knn)
```

```python
comparison = pd.DataFrame({
    'Method': ['SimpleImputer (Median)', 'KNNImputer'],
    'Age (Row 1)': [35.0, df_knn.loc[1, 'age']],
    'Income (Row 2)': [48500.0, df_knn.loc[2, 'income']]
})

print("Imputation Comparison:")
print(comparison.to_string(index=False))
```

```python
city_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

city_encoded = city_encoder.fit_transform(df[['city']])

city_df = pd.DataFrame(
    city_encoded,
    columns=city_encoder.get_feature_names_out(['city'])
)

print("OneHotEncoded City:")
print(city_df)
```

```python
size_encoder = OrdinalEncoder(
    categories=[['S', 'M', 'L']]
)

df['size_encoded'] = size_encoder.fit_transform(df[['size']])

print("DataFrame After Ordinal Encoding:")
print(df)
```

```python
df['target'] = ['Yes', 'No', 'Yes', 'No', 'Yes']

label_encoder = LabelEncoder()

df['target_encoded'] = label_encoder.fit_transform(df['target'])

print("DataFrame After Label Encoding:")
print(df)
```

```python
final_df = pd.concat(
    [
        df[['age', 'income']],
        city_df,
        df[['size_encoded', 'target_encoded']]
    ],
    axis=1
)

print("Final Encoded DataFrame:")
print(final_df)
```
