```python
# ============================================================
# WEEK 2 ASSESSMENT - TASK 2
# Preprocessing, Features and Pipeline
# ============================================================


# 1. IMPORT LIBRARIES

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

import joblib
from pathlib import Path

print("Libraries imported successfully!")


# ============================================================
# 2. CREATE SYNTHETIC LOAN DATASET
# ============================================================

np.random.seed(42)

n = 600

df = pd.DataFrame({

    "application_date": pd.date_range(
        start="2024-01-01",
        periods=n,
        freq="D"
    ).astype(str),

    "income": np.random.randint(20000, 150000, n),

    "loan_amount": np.random.randint(50000, 500000, n),

    "credit_score": np.random.randint(300, 850, n),

    "employment_type": np.random.choice(
        ["Salaried", "Self-employed", "Business"], n
    ),

    "education_level": np.random.choice(
        ["High School", "Diploma", "Bachelor", "Master", "PhD"], n
    ),

    "city": np.random.choice(
        ["Chennai", "Coimbatore", "Bangalore", "Erode"], n
    )
})


# Create target variable

risk_score = (
    -2.0 * (df["income"] / df["loan_amount"])
    - 0.002 * (df["credit_score"] - 300)
    + np.random.normal(0, 0.25, n)
)

threshold = np.quantile(risk_score, 0.55)

df["loan_default"] = (risk_score > threshold).astype(int)

print("\nDataset created successfully!")

print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 3. INTRODUCE MISSING VALUES
# ============================================================

for column, count in [
    ("income", 30),
    ("credit_score", 25),
    ("employment_type", 20),
    ("education_level", 15),
    ("city", 10)
]:

    missing_indices = np.random.choice(
        df.index, count, replace=False
    )

    df.loc[missing_indices, column] = np.nan


print("\nMissing Values:")

print(df.isnull().sum())


# ============================================================
# 4. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop("loan_default", axis=1)

y = df["loan_default"]

print("\nFeatures Shape:", X.shape)

print("Target Shape:", y.shape)

print("\nTarget Distribution:")

print(y.value_counts())


# ============================================================
# 5. TRY KNN IMPUTER ON NUMERIC COLUMNS
# ============================================================

numeric_cols = [
    "income",
    "loan_amount",
    "credit_score"
]

knn_imputer = KNNImputer(n_neighbors=5)

knn_result = knn_imputer.fit_transform(
    X[numeric_cols]
)

knn_result = pd.DataFrame(
    knn_result,
    columns=numeric_cols
)

print("\nKNN Imputation Results:")

print("Missing Values Before:")

print(X[numeric_cols].isnull().sum())

print("\nMissing Values After:")

print(knn_result.isnull().sum())

display(knn_result.head())


# ============================================================
# 6. FEATURE ENGINEERING
# ============================================================

class LoanFeatureEngineer(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        return self

    def transform(self, X):

        X = X.copy()

        # Date decomposition

        dates = pd.to_datetime(
            X["application_date"],
            errors="coerce"
        )

        X["application_month"] = dates.dt.month

        # Interaction feature

        X["income_loan_ratio"] = (
            X["income"] /
            X["loan_amount"].replace(0, np.nan)
        )

        # Log transformation

        X["log_loan_amount"] = np.log1p(
            X["loan_amount"].clip(lower=0)
        )

        # Remove original date column

        X = X.drop(columns=["application_date"])

        return X


feature_engineer = LoanFeatureEngineer()

engineered_data = feature_engineer.fit_transform(X)

print("\nFeature Engineering Completed!")

print("\nNew Features:")

print([
    "application_month",
    "income_loan_ratio",
    "log_loan_amount"
])

display(engineered_data.head())


# ============================================================
# 7. DEFINE COLUMN TRANSFORMER
# ============================================================

numeric_features = [
    "income",
    "loan_amount",
    "credit_score",
    "application_month",
    "income_loan_ratio",
    "log_loan_amount"
]


categorical_features = [
    "employment_type",
    "city"
]


ordinal_features = [
    "education_level"
]


# Numeric preprocessing

numeric_pipeline = Pipeline(steps=[

    ("imputer", SimpleImputer(strategy="median")),

    ("scaler", StandardScaler())

])


# Categorical preprocessing

categorical_pipeline = Pipeline(steps=[

    ("imputer", SimpleImputer(strategy="most_frequent")),

    ("encoder", OneHotEncoder(handle_unknown="ignore"))

])


# Ordinal preprocessing

ordinal_pipeline = Pipeline(steps=[

    ("imputer", SimpleImputer(strategy="most_frequent")),

    ("encoder", OrdinalEncoder(

        categories=[[
            "High School",
            "Diploma",
            "Bachelor",
            "Master",
            "PhD"
        ]],

        handle_unknown="use_encoded_value",

        unknown_value=-1

    ))

])


# Combine preprocessing

preprocessor = ColumnTransformer(transformers=[

    ("numeric", numeric_pipeline, numeric_features),

    ("categorical", categorical_pipeline, categorical_features),

    ("ordinal", ordinal_pipeline, ordinal_features)

])


print("\nColumnTransformer created successfully!")


# ============================================================
# 8. BUILD COMPLETE MACHINE LEARNING PIPELINE
# ============================================================

model_pipeline = Pipeline(steps=[

    ("feature_engineering", LoanFeatureEngineer()),

    ("preprocessor", preprocessor),

    ("model", LogisticRegression(max_iter=2000))

])


print("\nMachine Learning Pipeline Created!")

print(model_pipeline)


# ============================================================
# 9. CROSS VALIDATION
# ============================================================

cv_scores = cross_val_score(

    model_pipeline,

    X,

    y,

    cv=5,

    scoring="accuracy"

)


print("\nCross Validation Accuracy Scores:")

print(cv_scores)


print("\nAverage CV Accuracy:")

print(round(cv_scores.mean(), 4))


print("\nStandard Deviation:")

print(round(cv_scores.std(), 4))


# ============================================================
# 10. TRAIN FINAL PIPELINE
# ============================================================

model_pipeline.fit(X, y)

print("\nFinal Pipeline Training Completed!")


# ============================================================
# 11. SAVE FITTED PIPELINE USING JOBLIB
# ============================================================

task_folder = "Task_2_Preprocessing_Features_Pipeline"

current_dir = Path.cwd()

if current_dir.name == task_folder:

    output_dir = current_dir

elif (current_dir / "WEEK2 ASSESSMENT" / task_folder).exists():

    output_dir = current_dir / "WEEK2 ASSESSMENT" / task_folder

elif (current_dir.parent / "WEEK2 ASSESSMENT" / task_folder).exists():

    output_dir = current_dir.parent / "WEEK2 ASSESSMENT" / task_folder

else:

    output_dir = current_dir

output_dir.mkdir(parents=True, exist_ok=True)

model_path = output_dir / "loan_pipeline.joblib"

joblib.dump(model_pipeline, model_path)

print("\nPipeline Saved Successfully!")

print("Saved Location:", model_path.resolve())


# ============================================================
# 12. LOAD SAVED PIPELINE
# ============================================================

loaded_pipeline = joblib.load(model_path)

print("\nPipeline Loaded Successfully!")


# ============================================================
# 13. PREDICT THREE NEW RAW DATA ROWS
# ============================================================

new_data = pd.DataFrame({

    "application_date": [
        "2025-01-15",
        "2025-02-20",
        "2025-03-10"
    ],

    "income": [
        75000,
        np.nan,
        120000
    ],

    "loan_amount": [
        200000,
        150000,
        300000
    ],

    "credit_score": [
        720,
        650,
        np.nan
    ],

    "employment_type": [
        "Salaried",
        "Self-employed",
        np.nan
    ],

    "education_level": [
        "Master",
        "Bachelor",
        "PhD"
    ],

    "city": [
        "Chennai",
        "Coimbatore",
        "Bangalore"
    ]

})


predictions = loaded_pipeline.predict(new_data)


print("\nNew Raw Data:")

display(new_data)


print("\nPredictions:")

print(predictions)


print("\nPrediction Results:")

for i, prediction in enumerate(predictions, start=1):

    print(f"Application {i}: {prediction}")


# ============================================================
# 14. TASK COMPLETION SUMMARY
# ============================================================

print("\n" + "=" * 50)

print("TASK 2 COMPLETED SUCCESSFULLY!")

print("=" * 50)

print("1. SimpleImputer - Median and Most Frequent")

print("2. KNNImputer - Numeric Columns")

print("3. OneHotEncoder - Categorical Features")

print("4. OrdinalEncoder - Ordered Features")

print("5. Feature Engineering - Three New Features")

print("6. ColumnTransformer - Preprocessing")

print("7. Pipeline - Complete ML Workflow")

print("8. Cross Validation - 5 Folds")

print("9. Joblib - Save and Load Model")

print("10. Predictions - Three New Raw Rows")

print("=" * 50)
```
